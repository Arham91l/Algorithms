```python
import numpy as np

x = np.array([10, 12, 11, 13, 12, 14, 13])
X = x[:-1].reshape(-1,1)
Y = x[1:]
X_design = np.hstack([np.ones((len(X),1)), X])
beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y
c = beta[0]
phi = beta[1]
print("Intercept (c):", c)
print("AR coefficient (phi):", phi)
```

```python
import pandas as pd
from statsmodels.tsa.ar_model import AutoReg

data = [10, 12, 11, 13, 12, 14, 13]

model = AutoReg(data, lags=1)

model_fit = model.fit()

```

```python
print("Intercept:", model_fit.params[0])
print("AR coefficient:", model_fit.params[1])
```
