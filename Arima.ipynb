```python
import numpy as np
series = [10, 12, 11, 13, 12, 14, 13]
```

```python
diff = []
for i in range(1,len(series)):
    diff.append(series[i] - series[i-1])
diff
```

```python
def loss(phi, theta, diff):

    errors = [0]

    sse = 0

    for t in range(1, len(diff)):

        pred = (phi * diff[t-1] +theta * errors[-1])
        error = diff[t] - pred
        errors.append(error)
        sse += error**2
    return sse
```

```python
best_loss = float('inf')

best_phi = None
best_theta = None

for phi in np.arange(-1, 1.01, 0.01):

    for theta in np.arange(-1, 1.01, 0.01):

        current_loss = loss(
            phi,
            theta,
            diff
        )
        if current_loss < best_loss:

            best_loss = current_loss

            best_phi = phi

            best_theta = theta

print(best_phi)
print(best_theta)
print(best_loss)
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(series, order=(1,1,1))
fit = model.fit()
```

```python
ar_coef = fit.arparams
ma_coef = fit.maparams

print(ar_coef)
print(ma_coef)
```
