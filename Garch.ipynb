```python
import numpy as np

returns = np.array([
    0.01,-0.02,0.03,-0.01,
    0.04,-0.03,0.02
])
errors = returns - np.mean(returns)
variances = [np.var(errors)]
```

```python
best_loss = float('inf')

for omega in np.arange(0.0001,0.01,0.0005):

    for alpha in np.arange(0,1,0.05):

        for beta in np.arange(0,1,0.05):

            if alpha + beta >= 1:
                continue

            variances = [
                np.var(errors)
            ]

            loss = 0

            for t in range(1,len(errors)):

                pred = (
                    omega
                    +
                    alpha*errors[t-1]**2
                    +
                    beta*variances[-1]
                )

                variances.append(pred)

                actual = errors[t]**2

                loss += (
                    actual-pred
                )**2

            if loss < best_loss:

                best_loss = loss

                best_params = (
                    omega,
                    alpha,
                    beta
                )

print(best_params)
```
