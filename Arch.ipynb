```python
import numpy as np

returns = np.array([
    0.01,-0.02,0.03,-0.01,
    0.04,-0.03,0.02
])
errors = returns - np.mean(returns)
```

```python
best_loss = float('inf')
for omega in np.arange(0.0001,0.01,0.0001):
    for alpha in np.arange(0,1,0.01):

        loss = 0

        for t in range(1,len(errors)):

            pred_var = (omega + alpha*errors[t-1]**2)

            actual = errors[t]**2

            loss += (
                actual-pred_var
            )**2

        if loss < best_loss:

            best_loss = loss
            best_omega = omega
            best_alpha = alpha

print(best_omega)
print(best_alpha)
```
