## K-Nearest Neighbors (KNN)

KNN is a supervised learning algorithm used for classification and regression.

Unlike many machine learning algorithms, KNN does not learn explicit model parameters during training. Instead, it stores the training data and makes predictions based on the nearest neighbors of a query point.

### Working Principle

1. Choose a value of K.
2. Compute the distance between the query point and all training samples.
3. Select the K nearest neighbors.
4. Perform majority voting among the neighbors.
5. Assign the most common class as the prediction.

### Distance Metric

For this implementation, Euclidean Distance is used:



The class with the highest frequency among the K nearest neighbors is selected as the predicted class.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Class 0
X0 = np.random.randn(100, 2) + [0, 0]

# Class 1
X1 = np.random.randn(100, 2) + [5, 5]

# Class 2
X2 = np.random.randn(100, 2) + [0, 5]

X = np.vstack([X0, X1, X2])

y = np.array(
    [0]*100 +
    [1]*100 +
    [2]*100
)

df = pd.DataFrame(X, columns=["feature1", "feature2"])
df["target"] = y

print(df.head())
```

```python
query_point = np.array([[2.5, 3.0]])
distances = []
for i in range(len(X)):
    distance = np.sqrt(np.sum((X[i]-query_point)**2))
    distances.append(distance)
indices = np.argsort(distances)
k = 5
k_indices = indices[:k]
knn = y[k_indices]
print('Predicted Class:',np.bincount(knn).argmax())
print('Indices:',k_indices)
print('Distances:',np.sort(distances)[:k])
```

## K-Nearest Neighbors using Scikit-Learn

To verify our from-scratch implementation, we use Scikit-Learn's KNeighborsClassifier.

The model stores all training samples and predicts the class of a query point by:

1. Calculating the distance between the query point and all training samples.
2. Selecting the K nearest neighbors.
3. Performing majority voting among the selected neighbors.
4. Assigning the most frequent class as the prediction.

### Parameters

- Number of Neighbors (K): 5
- Distance Metric: Euclidean Distance

Scikit-Learn provides an efficient implementation of the KNN algorithm and serves as a benchmark for validating our custom implementation.

```python
from sklearn.neighbors import KNeighborsClassifier

# Train on full dataset
knn = KNeighborsClassifier(
    n_neighbors=5,
    metric='euclidean'
)

knn.fit(X, y)

query_point = np.array([[2.5, 3.0]])

# Get neighbors
distances, indices = knn.kneighbors(query_point)

print("Indices:", indices[0])
print("Distances:", distances[0])

# Prediction
prediction = knn.predict(query_point)

print("Predicted Class:", prediction[0])
```

## K-Nearest Neighbors Regression from Scratch

KNN Regression is a supervised learning algorithm used to predict continuous numerical values.

Unlike parametric models, KNN does not learn explicit parameters during training. Instead, it stores the dataset and makes predictions based on the nearest neighbors of a query point.

### Working Principle

1. Compute the distance between the query point and all training samples.
2. Select the K nearest neighbors.
3. Retrieve the target values of the selected neighbors.
4. Compute the average of these target values.
5. Return the average as the prediction.

### Distance Metric

Euclidean Distance is used to measure similarity between data points.

The prediction is calculated as the mean of the target values of the K nearest neighbors.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# 300 samples, 3 features
X = np.random.rand(300, 3)

# True relationship
y = (
    4 * X[:, 0]
    + 2 * X[:, 1]
    - 3 * X[:, 2]
    + 5
    + np.random.normal(0, 0.5, 300)
)

df = pd.DataFrame(
    X,
    columns=["feature1", "feature2", "feature3"]
)

df["target"] = y

print(df.head())
```

```python
query_point = np.array([[0.5, 0.2, 0.8]])
distances = []
for i in range(len(X)):
    distance = np.sqrt(np.sum((X[i]-query_point)**2))
    distances.append(distance)
indices = np.argsort(distances)
k = 5
k_indices = indices[:k]
knn = y[k_indices]
print('Predicted output:',np.mean(knn))
print('Indices:',k_indices)
print('Distances:',np.sort(distances)[:k])
```

## K-Nearest Neighbors Regressor using Scikit-Learn

To validate our custom implementation, we use Scikit-Learn's KNeighborsRegressor.

The model predicts a continuous target value by:

1. Calculating the distance between the query point and all samples.
2. Identifying the K nearest neighbors.
3. Collecting their target values.
4. Computing the average target value.

Scikit-Learn provides an optimized implementation and serves as a benchmark for comparing the correctness of our from-scratch solution.

```python
from sklearn.neighbors import KNeighborsRegressor

# Create model
knn = KNeighborsRegressor(
    n_neighbors=5,
    metric='euclidean'
)

# Fit on full dataset
knn.fit(X, y)

# Query point
query_point = np.array([[0.5, 0.2, 0.8]])

# Prediction
prediction = knn.predict(query_point)

print("Predicted Value:", prediction[0])

# Get neighbors
distances, indices = knn.kneighbors(query_point)

print("\nNeighbor Indices:")
print(indices[0])

print("\nNeighbor Distances:")
print(distances[0])
```

## Results and Comparison

The prediction obtained from the from-scratch implementation closely matches the prediction generated by Scikit-Learn.

This confirms that:

- Euclidean distance was computed correctly.
- The nearest neighbors were identified correctly.
- The averaging logic for regression was implemented correctly.
- The KNN Regressor behaves as expected.
