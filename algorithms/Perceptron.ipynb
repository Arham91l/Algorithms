# Perceptron from Scratch

The Perceptron is one of the earliest neural network models and serves as the foundation of modern deep learning.

It is a binary classification algorithm that learns a linear decision boundary by adjusting weights whenever a sample is misclassified.

### Model Equation

z = w₁x₁ + w₂x₂ + ... + wₙxₙ + b

### Activation Function

The Perceptron uses a Step Function:

- If z ≥ 0 → Predict 1
- If z < 0 → Predict 0

### Learning Rule

The weights are updated only when a prediction is incorrect.

Error = Actual Output − Predicted Output

Weight Update:

w ← w + η(error)x

Bias Update:

b ← b + η(error)

where η is the learning rate.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Class 0
X0 = np.random.randn(100, 2) + np.array([2, 2])

# Class 1
X1 = np.random.randn(100, 2) + np.array([6, 6])

X = np.vstack([X0, X1])

y = np.array(
    [0] * 100 +
    [1] * 100
)

df = pd.DataFrame(
    X,
    columns=["feature1", "feature2"]
)

df["target"] = y

print(df.head())
```

## Training Process

For each training sample:

1. Compute the weighted sum of inputs.
2. Apply the step activation function.
3. Compare the prediction with the true label.
4. Calculate the prediction error.
5. Update weights and bias if an error occurs.

The process is repeated for multiple epochs until the model converges or the maximum number of epochs is reached.

```python
w = np.zeros(X.shape[1])
b = 0
learning_rate = 0.01
epochs = 1000
for i in range(epochs):
  for idx, x_i in enumerate(X):
    z = np.dot(w,x_i) + b
    y_pred = 1 if z>=0 else 0
    error =  y[idx] - y_pred 
    w = w + learning_rate * error * x_i
    b = b + learning_rate * error
print(w)
print(b)
```

## Perceptron using Scikit-Learn

To validate the correctness of our implementation, we train a Perceptron model using Scikit-Learn.

The Scikit-Learn implementation automatically:

- Computes predictions using a linear decision boundary.
- Updates weights when misclassifications occur.
- Iteratively improves the model over multiple epochs.

The learned parameters and prediction accuracy are compared with those obtained from the from-scratch implementation.

```python
from sklearn.linear_model import Perceptron
# Create model
perceptron = Perceptron(
    max_iter=1000,
    eta0=0.01,
    random_state=42
)

# Train
perceptron.fit(X, y)

# Predict
y_pred = perceptron.predict(X)

# Parameters
print("\nWeights:")
print(perceptron.coef_)

print("\nBias:")
print(perceptron.intercept_)
```

## Results and Comparison

The from-scratch Perceptron successfully learns a decision boundary that separates the two classes.

The prediction accuracy is comparable to that obtained using Scikit-Learn, demonstrating that:

- The step activation function was implemented correctly.
- The perceptron learning rule was applied correctly.
- The weight and bias updates successfully reduced classification errors.

Although the learned weights may differ from Scikit-Learn's values, similar prediction performance indicates that the implementation is correct.
