# Support Vector Machine (SVM) from Scratch

Support Vector Machine (SVM) is a supervised learning algorithm used for classification.

The objective is to find the hyperplane that maximizes the margin between classes.

### Hyperplane

wᵀx + b = 0

### Margin

The margin is the distance between the decision boundary and the nearest data points.

These nearest points are called Support Vectors.

### Training Objective

SVM minimizes hinge loss while maximizing the margin between classes.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Class -1
X1 = np.random.randn(100, 2) + np.array([2, 2])

# Class +1
X2 = np.random.randn(100, 2) + np.array([6, 6])

X = np.vstack([X1, X2])

y = np.array(
    [-1] * 100 +
    [1] * 100
)

df = pd.DataFrame(
    X,
    columns=["feature1", "feature2"]
)

df["target"] = y

print(df.head())
print(df["target"].value_counts())
```

```python
b = 0
w = np.ones(X.shape[1])
```

```python
epochs = 5000
lr = 0.001
lambda_param = 0.01

for epoch in range(epochs):

    for idx, x_i in enumerate(X):

        condition = y[idx] * (np.dot(x_i, w) + b) >= 1

        if condition:

            dw = 2 * lambda_param * w
            db = 0

        else:

            dw = 2 * lambda_param * w - y[idx] * x_i
            db = -y[idx]

        w -= lr * dw
        b -= lr * db
print(w)
print(b)
```

```python
from sklearn.svm import SVC

model = SVC(kernel='linear')
model.fit(X, y)
```

## Support Vector Machine (SVM) from Scratch

Support Vector Machine is a supervised learning algorithm used for binary classification.

The objective is to find the hyperplane that maximizes the margin between classes.

### Hyperplane

wᵀx + b = 0

### Prediction Rule

ŷ = sign(wᵀx + b)

### Hinge Loss

L = max(0, 1 - y(wᵀx + b))

### Training Process

1. Initialize weights and bias.
2. Compute the margin condition.
3. Calculate gradients using hinge loss.
4. Update parameters using gradient descent.
5. Repeat until convergence.
