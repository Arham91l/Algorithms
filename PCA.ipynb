# Principal Component Analysis (PCA)

Principal Component Analysis (PCA) is a dimensionality reduction technique used to transform a dataset into a smaller set of uncorrelated variables called principal components.

### Objectives

1. Reduce the number of features.
2. Preserve as much information (variance) as possible.
3. Remove redundancy caused by correlated features.

### PCA Workflow

1. Standardize the data.
2. Compute the covariance matrix.
3. Calculate eigenvalues and eigenvectors.
4. Sort eigenvalues in descending order.
5. Select the top principal components.
6. Project the data onto the new feature space.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_samples = 300

x1 = np.random.normal(0, 1, n_samples)

# Correlated features
x2 = 2 * x1 + np.random.normal(0, 0.5, n_samples)
x3 = x1 - x2 + np.random.normal(0, 0.2, n_samples)

X = np.column_stack([x1, x2, x3])

df = pd.DataFrame(
    X,
    columns=["feature1", "feature2", "feature3"]
)

print(df.head())
print("Shape:", X.shape)
```

```python
X.mean(axis=0)
```

```python
X_mean = X - X.mean(axis=0)
```

```python
X_cov = np.dot(X_mean.T,X_mean)/(X.shape[0]-1)
```

```python
X_cov
```

```python
eigenvalues, eigenvectors = np.linalg.eig(X_cov)
```

```python
idx = np.argsort(eigenvalues)[::-1]
```

```python
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
```

```python
X_pca = np.dot(X_mean,eigenvectors[:,:2])
```

```python
X_pca[:5]
```

## Principal Component Analysis using Scikit-Learn

To validate the from-scratch implementation, we use Scikit-Learn's PCA module.

The algorithm:

1. Computes the covariance structure of the data.
2. Finds the principal components using eigen decomposition.
3. Sorts components according to explained variance.
4. Projects the data onto the selected principal components.

The transformed data, explained variance, and principal components are compared with the results obtained from the manual implementation.

```python
from sklearn.decomposition import PCA

# Reduce from 3 dimensions to 2 dimensions
pca = PCA(n_components=2)

X_pca_sklearn = pca.fit_transform(X_mean)

print("Original Shape:", X.shape)
print("Reduced Shape:", X_pca_sklearn.shape)

print("\nTransformed Data:")
print(X_pca_sklearn[:5])
```
