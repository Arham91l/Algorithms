```python
import numpy as np

data = np.array([10,12,11,13,12,14,13])

mu = np.mean(data)

best_theta = None
best_loss = float('inf')

for theta in np.arange(-1, 1.01, 0.01):

    errors = [0]

    sse = 0

    for t in range(1, len(data)):

        pred = mu + theta * errors[-1]

        error = data[t] - pred

        errors.append(error)

        sse += error**2

    if sse < best_loss:

        best_loss = sse
        best_theta = theta

print("Theta:", best_theta)
print("Loss:", best_loss)
```
