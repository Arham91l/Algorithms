# Linear Regression from Scratch

In this notebook, we implement Multivariate Linear Regression using Gradient Descent from scratch with NumPy and compare the results with Scikit-Learn's implementation.

The objective is to learn how the model updates weights and bias to minimize prediction error.

## Step 1: Generate Synthetic Dataset

We create a dataset with:
- 100 samples
- 3 input features
- A target variable generated using a known linear relationship

Target equation:

y = 4x₁ + 2x₂ - 3x₃ + 5 + noise

Random noise is added to simulate real-world data.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# 100 samples, 3 features
X = np.random.rand(100, 3)

# True weights
w = np.array([4, 2, -3])

# Generate target
y = X @ w + 5 + np.random.randn(100) * 0.5

# Convert to DataFrame for readability
df = pd.DataFrame(X, columns=['feature1', 'feature2', 'feature3'])
df['target'] = y

print(df.head())
```

## Step 2: Initialize Parameters

We initialize:
- Weights (m) for each feature
- Bias (b)

These parameters will be updated using Gradient Descent to minimize the Mean Squared Error (MSE) loss.

```python
b = 50
m = np.array([100,100,100])
```

## Step 3: Train Linear Regression using Gradient Descent

Training Process:

1. Compute predictions
2. Calculate gradients of the loss function
3. Update weights and bias
4. Repeat for a fixed number of epochs

Gradient formulas:

db = (-2/n) Σ(y - ŷ)

dm = (-2/n) X(y - ŷ)

where:
- n = number of samples
- y = actual values
- ŷ = predicted values

```python
learning_rate = 0.1
epochs = 1000
for i in range(epochs):
    y_hat = b + np.dot(m,X.T)
    db = (-2*np.sum(y-y_hat))/(X.shape[0])
    dm = (-2*np.dot((y-y_hat),X))/(X.shape[0])
    b = b - (learning_rate)*(db)
    m = m - (learning_rate)*(dm)
print(b)
print(m)
```

## Step 4: Compare with Scikit-Learn

To verify our implementation, we train a Linear Regression model using Scikit-Learn.

We compare:
- Learned weights
- Learned bias
- Prediction performance

The learned parameters should be close to the true values used to generate the dataset.

```python
from sklearn.linear_model import LinearRegression
# Train model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

# Learned parameters
print("Intercept (b):", model.intercept_)
print("Weights (m):", model.coef_)

```
